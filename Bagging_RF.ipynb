```python
# Import necessary libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.metrics import accuracy_score

# Load the Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=48)
```

```python

# -------------------------
# Decision Tree without Bagging
# -------------------------
# Create and train a single decision tree
dt_model = DecisionTreeClassifier(random_state=48)
dt_model.fit(X_train, y_train)
# is this based on the gini impurity or entropy? ---> gini impurity
# if we want to use entropy, we can specify it as follows:
# dt_model = DecisionTreeClassifier(criterion='entropy', random_state=42)

# Make predictions using the decision tree
dt_predictions = dt_model.predict(X_test)

# Evaluate accuracy of decision tree
dt_accuracy = accuracy_score(y_test, dt_predictions)
print(f"Accuracy of Decision Tree (no bagging): {dt_accuracy:.2f}")
```

```python
# -------------------------
# Decision Tree with Bagging
# -------------------------
# Create a Bagging classifier using Decision Tree as the base model
bagging_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(),  # Correct parameter name
    n_estimators=100,                     # Number of trees
    random_state=48,
    bootstrap=True                       # Sampling with replacement
)
# Train the Bagging classifier
bagging_model.fit(X_train, y_train)

# Make predictions using the bagging model
bagging_predictions = bagging_model.predict(X_test)

# Evaluate accuracy of Bagging classifier
bagging_accuracy = accuracy_score(y_test, bagging_predictions)
print(f"Accuracy of Decision Tree with Bagging: {bagging_accuracy:.2f}")
```

```python
# -------------------------
# Decision Tree with Random Forest
# -------------------------

from sklearn.ensemble import RandomForestClassifier

# Create a Random Forest classifier
rf_model = RandomForestClassifier(
   n_estimators=100,         # Number of trees
    criterion='gini',         # Splitting criterion (gini or entropy)
    max_depth=None,           # Grow trees until all leaves are pure
    random_state=48
)
# Train the Random Forest classifier
rf_model.fit(X_train, y_train)

# Make predictions using the Random Forest model
rf_predictions = rf_model.predict(X_test)

# Evaluate accuracy of Random Forest classifier
rf_accuracy = accuracy_score(y_test, rf_predictions)
print(f"Accuracy of Random Forest: {rf_accuracy:.2f}")

```
